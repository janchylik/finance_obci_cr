```python
import glob
import os

import pandas as pd

klicove_casti = {
    'ZC_ICO' : 'ico',
    'FISCPER' : 'obdobi_sap',
    'FUNC_AREA' : 'paragraf_id',
    'ZCMMT_ITM' : 'polozka_id',
    'ZU_ROZKZ' : 'skutecnost_czk'
}

soubory = glob.glob("../data/raw/FINM201*.csv")
print(f"Nalezeno souborů ke spojení: len{(soubory)}")

list_tabulek = []

for soubor in soubory:
    print(f"Zpracovávám: {os.path.basename(soubor)}")
    
    hlavicka = pd.read_csv(soubor, sep=';', nrows=0).columns
    
    mapa_na_miru = {}
    
    for skutecny_sloupec in hlavicka:
        for klic, cisty_nazev in klicove_casti.items():
            if klic in skutecny_sloupec:
                mapa_na_miru[skutecny_sloupec] = cisty_nazev
    
    df = pd.read_csv(soubor, sep=';', usecols=mapa_na_miru.keys(), low_memory=False)
    
    df = df.rename(columns=mapa_na_miru)
    
    list_tabulek.append(df)

kompletni_data = pd.concat(list_tabulek, ignore_index=True)
print(f"\nSpojeno! Celkový počet řádků k exportu: {len(kompletni_data): ,}")

cesta_vystup = "../data/processed/finm201_komplet.csv"

kompletni_data.to_csv(cesta_vystup, index=False, encoding='utf-8')
print(f"Úspěšně uloženo do: {cesta_vystup}")
```
